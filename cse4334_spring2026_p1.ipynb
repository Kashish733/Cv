````
import time
import os
from matplotlib import pyplot as plt, rc
from matplotlib.animation import FuncAnimation


# Your code goes inside the llps() function. You may add other functions, and
# you will likely need to do so. However, you must not modify any existing
# part of this skeleton file.
def llps(sequence):
  local_prominent_streaks = [[]] * len(sequence)
  growing_streaks = [[]] * len(sequence)
  prominent_streaks = []

  curr_growing = []

  def is_dominated(new_s):
    while prominent_streaks and \
     prominent_streaks[-1] != new_s and \
     (prominent_streaks[-1][1] - prominent_streaks[-1][0]) < (new_s[1] - new_s[0]) and \
     prominent_streaks[-1][2] <= new_s[2]
          prominent_streaks.pop()

    for a in prominent_streaks:
      if (a[1]-a[0]) >= (new_s[1] - new_s[0]) and a[2] >= new_s[2]:
        return
    prominent_streaks.append(new_s)

  for i in range(len(sequence)):
    v = sequence[i]
    local = []
    new_streak = []

    greater = None
    has_equal = False

    for s in curr_growing:
      if s[2] < v:
        new_streak.append([s[0], i+1, s[2]])
      elif s[2] == v:
        new_streak.append([s[0], i+1, s[2]])
        has_equal = True
      else:
        local.append(s)
        if greater is None:
          greater = s

    if has_equal:
      pass
    elif greater is not None:
      new_streak.append([greater[0], i+1, v])
    else:
      new_streak.append([i+1, i+1, v])

    if greater is not None:
      is_dominated(greater)

    local_prominent_streaks[i] = local
    growing_streaks[i] = new_streak
    curr_growing = new_streak

  for s in curr_growing:
    is_dominated(s)
  local_prominent_streaks[len(sequence) - 1] = local_prominent_streaks[len(sequence) - 1]


  return local_prominent_streaks, growing_streaks, prominent_streaks


# To use the hard-coded results, call the main functon by main(hardcode=True).
# The hard-coded results are provided to help you observe how the rest of
# the code works, including the animation. The results are for the file
# 'toy' which has the 10-number sequence we used in class:
# [3, 1, 7, 7, 2, 5, 4, 6, 7, 3].
def hardcode():
  sequence = [3, 1, 7, 7, 2, 5, 4, 6, 7, 3]

  local_prominent_streaks = [
    [],
    [[1, 1, 3]],
    [],
    [],
    [[3, 4, 7]],
    [],
    [[6, 6, 5]],
    [],
    [],
    [[6, 9, 4], [8, 9, 6], [9, 9, 7]]
  ]

  growing_streaks = [
    [[1, 1, 3]],
    [[1, 2, 1]],
    [[1, 3, 1], [3, 3, 7]],
    [[1, 4, 1], [3, 4, 7]],
    [[1, 5, 1], [3, 5, 2]],
    [[1, 6, 1], [3, 6, 2], [6, 6, 5]],
    [[1, 7, 1], [3, 7, 2], [6, 7, 4]],
    [[1, 8, 1], [3, 8, 2], [6, 8, 4], [8, 8, 6]],
    [[1, 9, 1], [3, 9, 2], [6, 9, 4], [8, 9, 6], [9, 9, 7]],
    [[1, 10, 1], [3, 10, 2], [6, 10, 3]]
  ]

  prominent_streaks = [
    [1, 10, 1], [3, 10, 2], [6, 10, 3], [6, 9, 4], [3, 4, 7]
  ]

  return sequence, local_prominent_streaks, growing_streaks, prominent_streaks


def init(sequence):
  # You shouldn't change this function.
  # The code is for setting up the animation.
  global ax
  global lines

  # Set up the drawing area
  ax = plt.axes(xlim=(0, len(sequence)+1), ylim=(0, max(sequence)+1))
  ax.set_xlabel('Left-end of streak interval')
  ax.set_ylabel('Minival value in the streak')

  # Draw an empty line since the init() function must return some lines
  lines, = ax.plot([], [], lw=2)
  return lines,


def animate(i, sequence,growing_streaks):
  # You shouldn't change this function.
  # The code is for setting up the animation.
  global ax
  global lines

  if i % 2 == 0:
    # Draw an horizontal line that represents the value at the
    # [i//2 + 1]-th position in the sequence.
    # If i equals 0, it is the 1st position in the sequence.
    # If i equals 2, it is the 2nd position in the sequence.
    # And so on.
    ax.axhline(y=sequence[i//2], color='red', linestyle='-.')
    lines.set_data([], [])
    return lines,
  else:
    # Draw the markers that represent the growing streaks after the
    # [i//2+1]-th value in the sequence.
    # If i equals 1, it is the 1st position in the sequence.
    # If i equals 3, it is the 2nd position in the sequence.
    # And so on.
    ax.clear()
    ax.set_xlabel('Left-end of streak interval')
    ax.set_ylabel('Minival value in the streak')
    ax.set_xlim(0, len(sequence)+1)
    ax.set_ylim(0, max(sequence)+1)
    for gs in growing_streaks[i//2]:
      sc, = ax.plot(gs[0], gs[2], marker='s', markerfacecolor='none', markeredgecolor='green', markersize=12, ls='')
    return sc,


def animate_llps(sequence, growing_streaks):
  # You shouldn't change this function.
  # The code is for setting up the animation.
  fig = plt.figure(figsize=(8,8))
  ax = None
  lines = None

  ani = FuncAnimation(fig, animate, init_func=lambda: init(sequence), frames=len(sequence*2), fargs=(sequence, growing_streaks), interval=1000, blit=True)
  rc('animation', html='html5')
  return ani # Return the animation object


def prepare_data(DATA_DIR, FOLDER_ID):
  # Install gdown if not already installed
  os.system('uv add gdown')
  import gdown

  os.makedirs(DATA_DIR, exist_ok=True)

  if not os.listdir(DATA_DIR):  # Only download if folder is empty
    gdown.download_folder(
        id=FOLDER_ID,
        output=DATA_DIR,
        quiet=False,
        use_cookies=False  # No auth needed for public folders
    )


def main(use_hardcode=True):
    ani = None

    # Public Google Drive folder ID for the data files
    FOLDER_ID = '1OM2LLa9D7k9f01rBpdDTwNtuQNvTZIUX'
    # Local (Colab works too) data file folder
    DATA_DIR = './data'

    # Download files to the data file folder
    prepare_data(DATA_DIR, FOLDER_ID)

    # The file 'toy' has the 10-number sequence we used in class:
    # [3, 1, 7, 7, 2, 5, 4, 6, 7, 3].
    # Colab will crash given the large data file 'WC98'. So make sure
    # to take out 'WC98' from the file list when you run it in Colab.

    files = ['toy', 'gold', 'sp500', 'aol']
    for file in files:
        with open(DATA_DIR + '/' + file, 'r') as filehandle:
            print(file)
            sequence = [int(current_place.rstrip()) for current_place in filehandle.readlines()]
            start_time = time.time()

            # You can use the hardcode() function to understand how the rest of
            # the code works for the 'toy' file. You will need to implement
            # llps() and test if it produces the same results as hardcode() for
            # this file. More importantly, you should test your implementation
            # of llps() with all provided data files and your own test cases if
            # necessary.

            if file == 'toy' and use_hardcode:
              # Use this line for getting hardcoded results.
              sequence, local_prominent_streaks, growing_streaks, prominent_streaks = hardcode()
            else:
              # Use this line for calling your implementaion of the llps() function.
              local_prominent_streaks, growing_streaks, prominent_streaks = llps(sequence)

            print("Total execution time = %f seconds" % (time.time() - start_time))
            print("Length of Sequence: ", len(sequence))
            print("Number of Local Prominent Streaks: ", sum(len(sublist) for sublist in local_prominent_streaks))
            print("Number of Still Growing Streaks: ", len(growing_streaks[-1]))
            print("Number of Prominent Streaks: ", len(prominent_streaks))
            print(prominent_streaks)
            print()

            # Animation of the LLPS algorithm is provided to help you understand
            # how the algorithm works. It only works well with the small values
            # and short sequence in the 'toy' file. You are not expected to make
            # the animation work well with other files.
            if file == 'toy':
              ani = animate_llps(sequence, growing_streaks)

    return ani


main(use_hardcode=False)
````
